```python
import pandas as pd

# Caminho base (ajuste se necessário)
path = "../data/03_transformed/tabelas_normalizadas/temas_normalizados.csv"

df = pd.read_csv(f"{path}")

# =========================
# TEMAS
# =========================
temas = (
    df["temas"]
    .dropna()
    .apply(lambda x: eval(x) if isinstance(x, str) else x)
    .explode()
    .dropna()
    .astype(str)
    .str.strip()
    .drop_duplicates()
    .sort_values()
)

temas_df = pd.DataFrame({"valor_original": temas})

temas_df.to_csv(
    "data/04_review/entrada_para_revisao/temas.csv",
    index=False
)

print(f"Temas gerados: {len(temas_df)}")
```
